```python
import numpy as np
import pandas as pd

# Load housing data
df = pd.read_csv('house_prices.csv')  
df.fillna({'price': 0, 'bedrooms': 0}, inplace=True)  # Handle NaN
```

```python
# Create a matrix from bedrooms and prices
matrix = np.array(df[['bedrooms', 'price']])  # 2D array: [bedrooms, price]

```

```python
# Filter for 3, 4, or 5 bedrooms
mask = np.isin(matrix[:, 0], [3, 4, 5])  # Select rows where bedrooms (column 0) is 3, 4, or 5
filtered_matrix = matrix[mask]  # Filtered matrix
```

```python
# Summarize prices by bedroom count
bedroom_counts = [3, 4, 5]
for beds in bedroom_counts:
    bed_mask = filtered_matrix[:, 0] == beds  # Mask for specific bedroom count
    prices = filtered_matrix[bed_mask, 1]  # Prices for that bedroom count
    print(f"\n{beds} Bedrooms:")
    print(f"  Total Price: ${np.sum(prices):}")
    print(f"  Average Price: ${np.mean(prices):.1f}")
    print(f"  Number of Houses: {len(prices)}")
```

```python
# Print first few rows of filtered matrix
np.set_printoptions(formatter={'float': '{:.0f}'.format})
matrix[:, 0] = matrix[:, 0].astype(np.int32)
print("\nFirst 5 rows of filtered matrix (bedrooms, price):")
print(filtered_matrix[:5])
```

```python
# Add price range per bedroom
print("\nPrice Range by Bedroom Count:")
for beds in [3, 4, 5]:
    bed_mask = filtered_matrix[:, 0] == beds
    prices = filtered_matrix[bed_mask, 1]
    print(f"{beds} Bedrooms: ${np.min(prices):.1f} - ${np.max(prices):.1f}")
```
